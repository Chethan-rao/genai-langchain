### Agent with Memory

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

model = create_agent(
    model="groq:qwen/qwen3-32b",
    checkpointer=InMemorySaver()
)
```

```python
config={"configurable": {"thread_id": "test1"}}
```

```python
from langchain.messages import HumanMessage, SystemMessage

response = model.invoke({"messages": [HumanMessage(content="Hi my name is chethan")]}, config=config)
print(response)
```

```python
response = model.invoke({"messages": [HumanMessage(content="Whats my name")]}, config=config)
print(response) 
```
