### LLM models integration through LangChain
There are 2 ways to initialize models using langchain. 
1. Using init_chat_model which is provider-agnostic
2. Using provider specific class (like ChatGoogleGenerativeAI for gemini model)

You still need to install langchain-openai explicitly if you want to use OpenAI models, even when using the provider-agnostic initialization approach.

```python
# 1. Using init_chat_model

from langchain.chat_models import init_chat_model

model = init_chat_model(model="google_genai:gemini-2.5-flash-lite")

response = model.invoke("Hello, how are you?")

print(response.content)
```

```python
# 2. Using provider specific class

from langchain_google_genai import ChatGoogleGenerativeAI 

model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

response = model.invoke("Hello, how are you?")

print(response.content)
```

### Streaming and Batch

### Streaming

Most models can stream their output content while it is being generated. Bu displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time

```python
chunks = model.stream("Write me a 100 words paragraph on AI")
for chunk in chunks:
    print(chunk.text, end="", flush=True)
```

### Batching

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel.

```python
responses = model.batch(
    [
        "How are you",
        "What is the capital of France?",
    ]
)

for response in responses:
    print(response.content)
```
